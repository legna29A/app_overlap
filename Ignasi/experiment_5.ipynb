 # Análisis de Visualización de Clusters vs Perpetrador / Víctima

    ---

    ## 📌 Descripción General
    Este script carga resultados de clustering para **víctimas** y **perpetradores**, define funciones para generar gráficos 3D interactivos con Plotly y genera tablas con el conteo y el porcentaje de cada estado (Perpetrador / Víctima) por *cluster*.

    - **Imports y rutas**  
      - `pandas`: manejo de DataFrames  
      - `os`: creación de carpetas  
      - `plotly.express`: visualización 3D interactiva  
      - Rutas a CSV de resultados de clustering y variables  

    - **Funciones principales**  
      - `visualize_cluster_perpetrator(df)`  
      - `visualize_cluster_victim(df)`  

    ---

    ## 🔍 Flujo de Trabajo

    1. **Lectura de datos**  
       ```python
       vic_cl_df    = pd.read_csv(path_cl_vic)
       per_cl_df    = pd.read_csv(path_cl_per)
       pred_var_df  = pd.read_csv(path_var)
       target_df    = pd.read_csv(path_target)
       ```

    2. **Definición de funciones**

       ### `visualize_cluster_perpetrator(df)`

       - Crea la carpeta `perpetrador_clustering` si no existe.  
       - Prepara el *DataFrame*:  
         - Convierte `PERPETRADOR` a entero.  
         - Crea la etiqueta `PERP_LABEL` ↔ `{0: "No Perpetrador", 1: "Perpetrador"}`.  
       - Gráfico 3D con `px.scatter_3d`:  
         - Ejes: `umap_1`, `umap_2`, `umap_3`.  
         - Color por `PERP_LABEL` (verde / rojo).  
         - Información en *hover*: número de cluster.  
         - Guarda el HTML en `perpetrador_clustering/umap_3d_perpetrador.html`.  
       - Calcula la tabla de **conteo** y **porcentaje** por `(cluster, PERPETRADOR)`.  
       - Devuelve el *DataFrame* ordenado.  

       ### `visualize_cluster_victim(df)`

       - Crea la carpeta `victim_clustering` si no existe.  
       - Prepara el *DataFrame*:  
         - Convierte `VÍCTIMA` a entero.  
         - Crea la etiqueta `V_LABEL` ↔ `{0: "No Víctima", 1: "Víctima"}`.  
       - Gráfico 3D con `px.scatter_3d`:  
         - Ejes: `umap_1`, `umap_2`, `umap_3`.  
         - Color por `V_LABEL` (verde / rojo).  
         - Información en *hover*: número de cluster.  
         - Guarda el HTML en `victim_clustering/umap_3d_victima.html`.  
       - Calcula la tabla de **conteo** y **porcentaje** por `(cluster, VÍCTIMA)`.  
       - Devuelve el *DataFrame* ordenado.  

    ---

    ## 🏃‍♂️ Ejecución y guardado de resultados
    ```python
    counts_per_df = visualize_cluster_perpetrator(vic_cl_df)
    counts_per_df.to_csv("perpetrador_clustering/cluster_perpetrators.csv", index=False)
    display(counts_per_df)

    counts_vict_df = visualize_cluster_victim(per_cl_df)
    counts_vict_df.to_csv("victim_clustering/cluster_victima.csv", index=False)
    display(counts_vict_df)
    ```

    ---

    ## 📈 Interpretación
    - Cada gráfico 3D muestra cómo se distribuyen víctimas y perpetradores en el espacio UMAP 3D.  
    - Las tablas finales permiten observar, por *cluster*, la proporción de sujetos en cada estado.  
    - Esta nueva información (etiqueta de *cluster*) puede enriquecer un clasificador supervisado, aportando contexto sobre la probabilidad de ser víctima o perpetrador según el *cluster* asignado.  

    ---

    ## 🏆 Resultados Generados
    - **HTML 3D interactivo** para perpetradores: `perpetrador_clustering/umap_3d_perpetrador.html`

````python
def visualize_cluster_perpetrator(df):
    """
    Genera un gráfico 3D interactivo de UMAP coloreado por PERPETRADOR,
    guarda el HTML en 'perpetrador_clustering' y devuelve un DataFrame
    con el conteo de pares (cluster, PERPETRADOR) y su porcentaje dentro de cada cluster.
    """
    # Crear carpeta si no existe
    output_dir = "perpetrador_clustering"
    os.makedirs(output_dir, exist_ok=True)
    
    # Preparar DataFrame
    df = df.copy()
    df['PERPETRADOR'] = df['PERPETRADOR'].astype(int)
    df['PERP_LABEL'] = df['PERPETRADOR'].map({
        0: 'No Perpetrador',
        1: 'Perpetrador'
    })
    
    # Crear gráfico 3D
    fig = px.scatter_3d(
        df,
        x='umap_1', y='umap_2', z='umap_3',
        color='PERP_LABEL',
        color_discrete_sequence=['green', 'red'],
        labels={'PERP_LABEL': 'Estado'},
        title='UMAP 3D – Perpetradores vs No Perpetradores',
        hover_data=['cluster']
    )
    fig.update_traces(marker=dict(size=4))
    fig.update_layout(
        legend=dict(title='Perpetrador', itemsizing='constant'),
        margin=dict(l=0, r=0, b=0, t=30)
    )

    # Mostrar en notebook
    fig.show()

    # Guardar HTML
    html_path = os.path.join(output_dir, "umap_3d_perpetrador.html")
    fig.write_html(html_path, include_plotlyjs='cdn')

    # Calcular conteos y porcentaje
    counts = (
        df[['cluster', 'PERPETRADOR']]
        .value_counts()
        .reset_index(name='count')
    )
    # Calcular total por cluster
    total_cluster = counts.groupby('cluster')['count'].transform('sum')
    counts['percentage'] = (counts['count'] / total_cluster * 100).round(2)
    
    # Ordenar y resetear índice
    counts = counts.sort_values(['cluster', 'PERPETRADOR']).reset_index(drop=True)
    
    return counts

def visualize_cluster_victim(df):
    """
    Genera un gráfico 3D interactivo de UMAP coloreado por VÍCTIMA,
    guarda el HTML en 'victim_clustering' y devuelve un DataFrame
    con el conteo de pares (cluster, VÍCTIMA) y su porcentaje dentro de cada cluster.
    """
    # Crear carpeta si no existe
    output_dir = "victim_clustering"
    os.makedirs(output_dir, exist_ok=True)
    
    # Preparar DataFrame
    df = df.copy()
    df['VÍCTIMA'] = df['VÍCTIMA'].astype(int)
    df['V_LABEL'] = df['VÍCTIMA'].map({
        0: 'No víctima',
        1: 'Víctima'
    })
    
    # Crear gráfico 3D
    fig = px.scatter_3d(
        df,
        x='umap_1', y='umap_2', z='umap_3',
        color='V_LABEL',
        color_discrete_sequence=['green', 'red'],
        labels={'V_LABEL': 'Estado'},
        title='UMAP 3D – Víctima vs No Víctima',
        hover_data=['cluster']
    )
    fig.update_traces(marker=dict(size=4))
    fig.update_layout(
        legend=dict(title='Víctima', itemsizing='constant'),
        margin=dict(l=0, r=0, b=0, t=30)
    )

    # Mostrar en notebook
    fig.show()

    # Guardar HTML
    html_path = os.path.join(output_dir, "umap_3d_victima.html")
    fig.write_html(html_path, include_plotlyjs='cdn')

    # Calcular conteos y porcentaje
    counts = (
        df[['cluster', 'VÍCTIMA']]
        .value_counts()
        .reset_index(name='count')
    )
    # Calcular total por cluster
    total_cluster = counts.groupby('cluster')['count'].transform('sum')
    counts['percentage'] = (counts['count'] / total_cluster * 100).round(2)
    
    # Ordenar y resetear índice
    counts = counts.sort_values(['cluster', 'VÍCTIMA']).reset_index(drop=True)
    
    return counts
````

````python
# Clústeres hechos en función del típo de víctima.
# Vemos rápidamente que tipo de víctima tiene más probabilidad de ser perpetrador en base a su frecuencia dentro del clúster. 
# Vemos también que esta nueva variable (cluster) nos puede ayudar mucho a enriquecer al clasificador de perpretador. 
# Es decir, sabiendo el tipo de víctima junto con las variables globales seguramente la performance del clasificador aumentará.
# Asumción: Sabemos que tipo de víctima es -> Saber si es perpretador o no.   
counts_per_df = visualize_cluster_perpetrator(df_vic_cl_target)
counts_per_df.to_csv(r"perpetrador_clustering\cluster_perpetrators.csv",index=False)
display(counts_per_df)
````
