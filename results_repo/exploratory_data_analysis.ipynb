```python
def filter_strings(strings):
    """
    Splits the given list of strings into two lists:
    - valid: strings with no dots ('.'), no underscores ('_'), and no numeric digits.
    - invalid: all other strings.

    Args:
        strings (List[str]): List of input strings.

    Returns:
        Tuple[List[str], List[str]]: A tuple containing two lists (valid, invalid).
    """
    # Compile a regex to detect dots, underscores, or digits
    pattern = re.compile(r"[._\d]")
    valid = []
    invalid = []

    for s in strings:
        # If any forbidden character is found, classify as invalid
        if pattern.search(s):
            invalid.append(s)
        else:
            valid.append(s)

    return valid, invalid
```

```python
def create_main_prefix_dict(unique_prefixes, string_list):
    """
    Creates a dictionary where keys are unique prefixes and values are all strings
    that start with the prefix followed by a dot.

    Args:
        unique_prefixes (list): List of unique string prefixes
        string_list (list): List of strings to search through

    Returns:
        dict: Dictionary mapping each prefix to matching strings
    """
    prefix_dict = defaultdict(list)

    for prefix in unique_prefixes:
        for s in string_list:
            if s.startswith(prefix):
                prefix_dict[prefix].append(s)

    return dict(prefix_dict)
```

```python
#Funciones para visualizar proporciones.
def graficar_proporcion_por_categoria(df, filter_col, binary_col):
    """
    Crea una figura 2x2 con barplots de la proporción de una variable binaria
    para cada categoría de la columna especificada en filter_col, añadiendo el conteo encima de cada barra.

    Parámetros:
    -----------
    df : pandas.DataFrame
        DataFrame que contiene las columnas de interés.
    filter_col : str
        Nombre de la columna categórica por la que filtrar.
    binary_col : str
        Nombre de la columna binaria cuyos valores mostrar (debe contener 2 categorías).
    """
    # Obtener valores únicos de filter_col y asegurar orden consistente
    categorias = sorted(df[filter_col].dropna().unique())

    # Determinar tamaño de la cuadrícula (2x2 por defecto, adaptable si >4 categorías)
    n = len(categorias)
    cols = 2
    rows = (n + 1) // 2

    # Crear figura y ejes
    fig, axes = plt.subplots(rows, cols, figsize=(12, 4 * rows))
    axes = axes.flatten()

    # Iterar sobre cada categoría y su correspondiente subplot
    for ax, cat in zip(axes, categorias):
        sub_df = df[df[filter_col] == cat]
        # Calcular conteo y proporción de cada valor de binary_col
        conteos = sub_df[binary_col].value_counts().sort_index()
        proporciones = conteos / conteos.sum()

        # Graficar barras
        barras = ax.bar(
            conteos.index.astype(str),
            proporciones.values
        )

        # Añadir conteo encima de cada barra
        for bar, count in zip(barras, conteos.values):
            altura = bar.get_height()
            ax.text(
                bar.get_x() + bar.get_width() / 2,
                altura + 0.02,
                str(count),
                ha='center',
                va='bottom'
            )

        ax.set_title(f'{filter_col} = {cat}')
        ax.set_xlabel(binary_col)
        ax.set_ylabel('Proporción')
        ax.set_ylim(0, 1.1)

    # Ocultar ejes extra si hay más subplots que categorías
    for ax in axes[n:]:
        ax.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
# Función para el estudio de correlación entre categorías.
def cramer_v_categorica_binaria(df, col_cat, col_bin):
    """
    Calcula la asociación entre una variable categórica (col_cat)
    y una variable binaria (col_bin) usando el estadístico V de Cramér.

    Parámetros:
    -----------
    df : pandas.DataFrame
        DataFrame que contiene las columnas de interés.
    col_cat : str
        Nombre de la columna categórica.
    col_bin : str
        Nombre de la columna binaria (debe contener exactamente dos valores únicos).

    Devuelve:
    ---------
    float
        Valor de V de Cramér entre 0 y 1.
    """
    # Construir tabla de contingencia
    tabla = pd.crosstab(df[col_cat], df[col_bin])

    # Calcular chi-cuadrado
    chi2, _, _, _ = chi2_contingency(tabla, correction=False)
    n = tabla.values.sum()
    # Dimensiones de la tabla
    k = min(tabla.shape)  # número de categorías de la variable con menos clases

    # Cálculo de V de Cramér
    v = np.sqrt(chi2 / (n * (k - 1)))
    return v
```

```python
from typing import List

def extraer_prefijos_letras(strings: List[str]) -> List[str]:
    """
    Dada una lista de strings, devuelve una nueva lista con el
    prefijo de cada string que contiene solo caracteres alfabéticos
    (a–z, A–Z), deteniéndose justo antes del primer carácter
    que no sea letra.
    """
    prefijos = []
    for s in strings:
        prefix = []
        for c in s:
            if c.isalpha():
                prefix.append(c)
            else:
                break
        prefijos.append(''.join(prefix))
    return prefijos

def sort_by_string_length(list_strings):
  """retorna la lista ordenada por longitud descendiente"""
  return sorted(list_strings, key=len, reverse=True)
```

```python
def transform_GENEROBIN_ORIENTSEXBN(df: pd.DataFrame) -> pd.DataFrame:
    """
    Imputa valores faltantes en las columnas 'GENERO_BIN' y 'ORIENTSEX.BN'.

    Reglas de imputación:
      - En 'GENERO_BIN', reemplaza NaN con 2.
      - En 'ORIENTSEX.BN', reemplaza NaN con 3.

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame que debe contener las columnas 'GENERO_BIN' y 'ORIENTSEX.BN'.

    Retorna
    -------
    pd.DataFrame
        Copia del DataFrame original con las dos columnas imputadas y convertidas a entero.

    Lanza
    -----
    KeyError
        Si falta alguna de las columnas requeridas.
    """
    # Copiamos el DataFrame para no modificar el original
    df_transformed = df.copy()

    required_cols = ["GENERO_BIN", "ORIENTSEX.BN"]
    missing = [col for col in required_cols if col not in df_transformed.columns]
    if missing:
        raise KeyError(f"Faltan las columnas requeridas: {missing}")

    # Imputación
    df_transformed["GENERO_BIN"]    = df_transformed["GENERO_BIN"].fillna(2).astype(int)
    df_transformed["ORIENTSEX.BN"] = df_transformed["ORIENTSEX.BN"].fillna(3).astype(int)

    return df_transformed

def transform_PPBULL(df: pd.DataFrame) -> pd.DataFrame:
    """
    Añade la columna 'PP.BULL' codificada según la presencia de bullying en PP1.BULL o PP2.BULL:

      - 1: si alguna de las columnas 'PP1.BULL' o 'PP2.BULL' es igual a 1.
      - 0: en cualquier otro caso (incluye valores 0 o NaN en ambas columnas).

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame que debe contener las columnas 'PP1.BULL' y 'PP2.BULL'.

    Retorna
    -------
    pd.DataFrame
        Copia del DataFrame original con la nueva columna 'PP.BULL'.

    Lanza
    -----
    KeyError
        Si faltan las columnas requeridas.
    """
    # Trabajar sobre copia para no modificar el original
    df_transformed = df.copy()

    required = ["PP1.BULL", "PP2.BULL"]
    missing = [col for col in required if col not in df_transformed.columns]
    if missing:
        raise KeyError(f"Faltan las columnas requeridas: {missing}")

    # Crear PP.BULL: 1 si alguna es 1, sino 0 (NaN → False)
    df_transformed["PP.BULL"] = (
        df_transformed[required]
        .eq(1)
        .any(axis=1)
        .astype(int)
    )

    return df_transformed

def transform_ABUSOSUBS(df: pd.DataFrame) -> pd.DataFrame:
    """
    Transforma las columnas 'ABUSOSUBS1' y 'ABUSOSUBS2' de la siguiente manera:

      1. Convierte todos los valores iguales a 1 en 0.
      2. Imputa los valores faltantes (NaN) con 1 una vez aplicada la conversión anterior.
      3. Asegura que el tipo final de la columna sea entero (int64).

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame de entrada que debe contener las columnas
        'ABUSOSUBS1' y 'ABUSOSUBS2'.

    Devuelve
    -------
    pd.DataFrame
        Una copia del DataFrame original con las columnas transformadas.

    Lanza
    -----
    KeyError
        Si faltan una o ambas columnas requeridas en el DataFrame.
    """
    # Trabajamos sobre una copia para no mutar el df original
    df_transformed = df.copy()
    cols = ["ABUSOSUBS1", "ABUSOSUBS2"]

    # Verificar que existan las columnas necesarias
    missing = [c for c in cols if c not in df_transformed.columns]
    if missing:
        raise KeyError(f"Faltan las columnas requeridas: {missing}")

    for col in cols:
        # 1) Convertir 1 → 0
        df_transformed[col] = df_transformed[col].replace(1, 0)
        # 2) Rellenar NaN → 1
        df_transformed[col] = df_transformed[col].fillna(1)
        # 3) Convertir a entero
        df_transformed[col] = df_transformed[col].astype(int)

    return df_transformed

def transform_AUTOEFIC(df: pd.DataFrame) -> pd.DataFrame:
    """
    Imputa valores faltantes en las columnas 'AUTOEFIC1' a 'AUTOEFIC5' con la media de cada columna,
    y añade dos nuevas columnas:
      - 'AUTOEFIC.MEAN': media por fila de los cinco ítems imputados.
      - 'AUTOEFIC.VAR' : varianza por fila de los cinco ítems imputados (varianza poblacional, ddof=0).

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame de entrada que debe contener las columnas
        'AUTOEFIC1', 'AUTOEFIC2', 'AUTOEFIC3', 'AUTOEFIC4', 'AUTOEFIC5'.

    Retorna
    -------
    pd.DataFrame
        Copia del DataFrame original con:
        - Las columnas 'AUTOEFIC1'...'AUTOEFIC5' imputadas sin NaN.
        - Dos columnas nuevas: 'AUTOEFIC.MEAN' y 'AUTOEFIC.VAR'.

    Lanza
    -----
    KeyError
        Si falta alguna de las columnas requeridas.
    """
    # Hacemos una copia para no modificar el df original
    df_transformed = df.copy()
    cols = [f"AUTOEFIC{i}" for i in range(1, 6)]

    # Verificar existencia de columnas
    missing = [c for c in cols if c not in df_transformed.columns]
    if missing:
        raise KeyError(f"Faltan las columnas requeridas: {missing}")

    # Imputación: llenar NaN con la media de cada columna
    for col in cols:
        mean_val = df_transformed[col].mean(skipna=True)
        df_transformed[col] = df_transformed[col].fillna(mean_val)

    # Cálculo de la media y varianza por fila
    df_transformed["AUTOEFIC.MEAN"] = df_transformed[cols].mean(axis=1)
    df_transformed["AUTOEFIC.VAR"]  = df_transformed[cols].var(axis=1, ddof=0)

    return df_transformed

def transform_CONVIVEN(df: pd.DataFrame) -> pd.DataFrame:
    """
    Ajusta la columna 'CONVIVEN.6' sumándole el valor de 'CONVIVEN.7' y luego
    capea todos los valores resultantes a un máximo de 1.

    Pasos:
      1. Verifica que existan las columnas 'CONVIVEN.1' … 'CONVIVEN.7'.
      2. Suma, para cada fila, CONVIVEN.6 + CONVIVEN.7 (tratando NaN como 0).
      3. Si el resultado de la suma es mayor que 1, lo deja en 1.
      4. Convierte la columna resultante a entero.

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame de entrada con las columnas 'CONVIVEN.1' … 'CONVIVEN.7'.

    Devuelve
    -------
    pd.DataFrame
        Copia del DataFrame con 'CONVIVEN.6' actualizada y capada a 1.
        Las demás columnas permanecen iguales.

    Lanza
    -----
    KeyError
        Si falta alguna de las columnas necesarias.
    """
    df_transformed = df.copy()
    expected_cols = [f"CONVIVEN.{i}" for i in range(1, 8)]
    missing = [c for c in expected_cols if c not in df_transformed.columns]
    if missing:
        raise KeyError(f"Faltan las columnas requeridas: {missing}")

    # Sumar CONVIVEN.6 + CONVIVEN.7 (NaN → 0), capear a 1, y convertir a entero
    summed = (
        df_transformed["CONVIVEN.6"].fillna(0) +
        df_transformed["CONVIVEN.7"].fillna(0)
    )
    df_transformed["CONVIVEN.6"] = summed.clip(upper=1).astype(int)

    return df_transformed

def transform_IMPULS(df: pd.DataFrame) -> pd.DataFrame:
    """
    Imputa los valores faltantes en las columnas de impulsividad y añade tres
    estadísticas por fila: media, varianza y mediana.

    Pasos:
      1. Verifica que existan las columnas:
         'IMPULS1', 'IMPULS2_REV', 'IMPULS3.REV', 'IMPULS4',
         'IMPULS5', 'IMPULS6', 'IMPULS7.REV', 'IMPULS8.REV'.
      2. Para cada una, imputa los NaN con la media de su propia columna.
      3. Crea una nueva columna 'IMPULS.MEAN' con la media por fila de las 8 variables.
      4. Crea una nueva columna 'IMPULS.VAR' con la varianza poblacional (ddof=0) por fila.
      5. Crea una nueva columna 'IMPULS.MEDIAN' con la mediana por fila.

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame que contiene las columnas de impulsividad.

    Retorna
    -------
    pd.DataFrame
        Copia del DataFrame original con las columnas imputadas y las tres
        estadísticas añadidas.

    Lanza
    -----
    KeyError
        Si faltan columnas requeridas.
    """
    # Hacemos una copia para no mutar el original
    df_transformed = df.copy()

    cols = [
        "IMPULS1", "IMPULS2_REV", "IMPULS3.REV", "IMPULS4",
        "IMPULS5", "IMPULS6", "IMPULS7.REV", "IMPULS8.REV"
    ]

    # Verificación de columnas
    missing = [c for c in cols if c not in df_transformed.columns]
    if missing:
        raise KeyError(f"Faltan las columnas requeridas: {missing}")

    # Imputación de NaN con la media de cada columna
    for col in cols:
        mean_val = df_transformed[col].mean(skipna=True)
        df_transformed[col] = df_transformed[col].fillna(mean_val)

    # Cálculo de estadísticas por fila
    df_transformed["IMPULS.MEAN"]   = df_transformed[cols].mean(axis=1)
    df_transformed["IMPULS.VAR"]    = df_transformed[cols].var(axis=1, ddof=0)
    df_transformed["IMPULS.MEDIAN"] = df_transformed[cols].median(axis=1)

    return df_transformed

def transform_APOYO(df: pd.DataFrame) -> pd.DataFrame:
    """
    Imputa valores faltantes en las columnas de apoyo y añade tres
    estadísticas por fila: media, varianza y mediana.

    Pasos:
      1. Verifica que existan las columnas:
         'APOYO1', 'APOYO2', 'APOYO3', 'APOYO4',
         'APOYO5', 'APOYO6', 'APOYO7'.
      2. Para cada una, imputa los NaN con la media de su propia columna.
      3. Crea una nueva columna 'APOYO.MEAN' con la media por fila de las 7 variables.
      4. Crea una nueva columna 'APOYO.VAR' con la varianza poblacional (ddof=0) por fila.
      5. Crea una nueva columna 'APOYO.MEDIAN' con la mediana por fila.

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame que contiene las columnas de apoyo.

    Retorna
    -------
    pd.DataFrame
        Copia del DataFrame original con las columnas imputadas y las tres
        estadísticas añadidas.

    Lanza
    -----
    KeyError
        Si faltan columnas requeridas.
    """
    # Hacemos una copia para no mutar el original
    df_transformed = df.copy()

    cols = [f"APOYO{i}" for i in range(1, 8)]

    # Verificación de columnas
    missing = [c for c in cols if c not in df_transformed.columns]
    if missing:
        raise KeyError(f"Faltan las columnas requeridas: {missing}")

    # Imputación de NaN con la media de cada columna
    for col in cols:
        mean_val = df_transformed[col].mean(skipna=True)
        df_transformed[col] = df_transformed[col].fillna(mean_val)

    # Cálculo de estadísticas por fila
    df_transformed["APOYO.MEAN"]   = df_transformed[cols].mean(axis=1)
    df_transformed["APOYO.VAR"]    = df_transformed[cols].var(axis=1, ddof=0)
    df_transformed["APOYO.MEDIAN"] = df_transformed[cols].median(axis=1)

    return df_transformed

def transform_VEXPQUIEN(df: pd.DataFrame) -> pd.DataFrame:
    """
    Crea una nueva columna 'VEXP.QUIEN' codificada según estas reglas:

      - 1: desconocido  (alguna de las columnas originales contiene un 1)
      - 0: conocido      (ninguna contiene un 1, y al menos un valor no es NaN)
      - 2: nadie         (todas las columnas originales son NaN)

    Pasos:
      1. Verifica que existan las columnas
         'VEXP1.QUIEN.1', 'VEXP2.QUIEN.1' y 'VEXP3.QUIEN.1'.
      2. Para cada fila, comprueba:
         a) Si todas las columnas son NaN → asigna 2.
         b) Si alguna columna == 1 → asigna 1.
         c) En caso contrario → asigna 0.

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame que debe contener las columnas
        'VEXP1.QUIEN.1', 'VEXP2.QUIEN.1', 'VEXP3.QUIEN.1'.

    Retorna
    -------
    pd.DataFrame
        Copia del DataFrame original con la nueva columna 'VEXP.QUIEN'.

    Lanza
    -----
    KeyError
        Si falta alguna de las columnas requeridas.
    """
    df_transformed = df.copy()
    cols = ["VEXP1.QUIEN.1", "VEXP2.QUIEN.1", "VEXP3.QUIEN.1"]

    # Validación de columnas
    missing = [c for c in cols if c not in df_transformed.columns]
    if missing:
        raise KeyError(f"Faltan las columnas requeridas: {missing}")

    # Identificar casos
    all_nan    = df_transformed[cols].isna().all(axis=1)
    has_one    = df_transformed[cols].eq(1).any(axis=1)

    # Asignar por defecto 'conocido' (0)
    df_transformed["VEXP.QUIEN"] = 0
    # Luego sobreescribir desconocido (1)
    df_transformed.loc[has_one, "VEXP.QUIEN"] = 1
    # Finalmente, nadie (2) donde todas son NaN
    df_transformed.loc[all_nan, "VEXP.QUIEN"] = 2

    return df_transformed

def transform_VEXPCONTACT(df: pd.DataFrame) -> pd.DataFrame:
    """
    Crea la columna 'VEXP.CONTACT' con codificación según el tipo de contacto,
    a partir de las columnas binarias de contacto en persona y virtual para
    VEXP1, VEXP2 y VEXP3.

    Codificación de 'VEXP.CONTACT':
      - 3 : nadie (todas las columnas de contacto son NaN)
      - 0 : presencial (al menos un 1 en las columnas de contacto presenciales)
      - 1 : virtual   (al menos un 1 en las columnas de contacto virtuales)
      - 2 : ambos      (al menos un 1 en presenciales y al menos un 1 en virtuales)

    Columnas de entrada esperadas:
      Presenciales:
        VEXP1.CONTACT.1  VEXP1.CONTACT.2  VEXP1.CONTACT.3
        VEXP2.CONTACT.1  VEXP2.CONTACT.2  VEXP2.CONTACT.3
        VEXP3.CONTACT.1  VEXP3.CONTACT.2  VEXP3.CONTACT.3

      Virtuales:
        VEXP1.CONTACT.4  ...  VEXP1.CONTACT.12
        VEXP2.CONTACT.4  ...  VEXP2.CONTACT.12
        VEXP3.CONTACT.4  ...  VEXP3.CONTACT.12

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame que contiene todas las columnas 'VEXP{1,2,3}.CONTACT.{1..12}'.

    Retorna
    -------
    pd.DataFrame
        Copia del DataFrame original con la columna adicional 'VEXP.CONTACT'.

    Lanza
    -----
    KeyError
        Si falta alguna de las columnas requeridas.
    """
    df_transformed = df.copy()

    # Generar listas de nombres de columna
    pres_cols = [
        f"VEXP{i}.CONTACT.{j}"
        for i in range(1, 4)
        for j in range(1, 4)
    ]
    virt_cols = [
        f"VEXP{i}.CONTACT.{j}"
        for i in range(1, 4)
        for j in range(4, 13)
    ]
    all_cols = pres_cols + virt_cols

    # Validar que existan todas las columnas
    missing = [c for c in all_cols if c not in df_transformed.columns]
    if missing:
        raise KeyError(f"Faltan las columnas requeridas: {missing}")

    # Identificar patrones por fila
    all_nan    = df_transformed[all_cols].isna().all(axis=1)
    has_pres   = df_transformed[pres_cols].eq(1).any(axis=1)
    has_virt   = df_transformed[virt_cols].eq(1).any(axis=1)

    # Inicializar con 0 (presencial)
    df_transformed["VEXP.CONTACT"] = 0

    # Ambos (virtual + presencial)
    df_transformed.loc[has_pres & has_virt, "VEXP.CONTACT"] = 2
    # Virtual únicamente
    df_transformed.loc[~has_pres & has_virt, "VEXP.CONTACT"] = 1
    # Nadie (todas NaN)
    df_transformed.loc[all_nan, "VEXP.CONTACT"] = 3

    return df_transformed

def transform_VMQUIEN(df: pd.DataFrame) -> pd.DataFrame:
    """
    Crea la columna 'VM.QUIEN' codificada según la presencia de respuestas
    en las columnas VM{1..4}.QUIEN.{1..5}.

    Codificación de 'VM.QUIEN':
      - 1 : padres          (alguna de VM{i}.QUIEN.1 o VM{i}.QUIEN.2 es 1)
      - 0 : no padres       (alguna de VM{i}.QUIEN.3, VM{i}.QUIEN.4 o VM{i}.QUIEN.5 es 1,
                             y ninguna de las de 'padres' es 1)
      - 2 : ambos           (al menos un 1 en grupo 'padres' y al menos un 1 en grupo 'no padres')
      - 3 : nadie (todas NaN en las 20 columnas)

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame que debe contener las columnas:
          VM1.QUIEN.1 … VM1.QUIEN.5,
          VM2.QUIEN.1 … VM2.QUIEN.5,
          VM3.QUIEN.1 … VM3.QUIEN.5,
          VM4.QUIEN.1 … VM4.QUIEN.5.

    Retorna
    -------
    pd.DataFrame
        Copia del DataFrame original con la nueva columna 'VM.QUIEN'.

    Lanza
    -----
    KeyError
        Si falta alguna de las columnas requeridas.
    """
    df_transformed = df.copy()

    # Columnas de "padres"
    parent_cols = [
        f"VM{i}.QUIEN.{j}"
        for i in range(1, 5)
        for j in (1, 2)
    ]
    # Columnas de "no padres"
    non_parent_cols = [
        f"VM{i}.QUIEN.{j}"
        for i in range(1, 5)
        for j in (3, 4, 5)
    ]
    all_cols = parent_cols + non_parent_cols

    # Validar existencia de todas las columnas
    missing = [c for c in all_cols if c not in df_transformed.columns]
    if missing:
        raise KeyError(f"Faltan las columnas requeridas: {missing}")

    # Flags por fila
    all_nan       = df_transformed[all_cols].isna().all(axis=1)
    has_parent    = df_transformed[parent_cols].eq(1).any(axis=1)
    has_non_par   = df_transformed[non_parent_cols].eq(1).any(axis=1)

    # Inicializamos con 0 ("no padres")
    df_transformed["VM.QUIEN"] = 0

    # 2 = ambos
    df_transformed.loc[has_parent & has_non_par, "VM.QUIEN"] = 2
    # 1 = padres solamente
    df_transformed.loc[has_parent & ~has_non_par, "VM.QUIEN"] = 1
    # 0 = no padres solamente (ya es 0 por defecto, se mantiene)
    # 3 = nadie (todas NaN)
    df_transformed.loc[all_nan, "VM.QUIEN"] = 3

    return df_transformed

def transform_WQUIEN(df: pd.DataFrame) -> pd.DataFrame:
    """
    Añade la columna 'W.QUIEN' codificada según respuestas en las columnas
    W{1..4}.QUIEN.{1..4}.

    Codificación de 'W.QUIEN':
      - 3 : Nadie (todas las columnas son NaN)
      - 1 : Padres (al menos un 1 en las preguntas .1 o .2 de cualquiera de W1–W4,
                  y ninguno en .3 o .4)
      - 0 : No padres (al menos un 1 en las preguntas .3 o .4 de cualquiera de W1–W4,
                      y ninguno en .1 o .2)
      - 2 : Ambos (al menos un 1 en .1/.2 y al menos un 1 en .3/.4)

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame que contiene las siguientes columnas:
          W1.QUIEN.1  W1.QUIEN.2  W1.QUIEN.3  W1.QUIEN.4
          W2.QUIEN.1  W2.QUIEN.2  W2.QUIEN.3  W2.QUIEN.4
          W3.QUIEN.1  W3.QUIEN.2  W3.QUIEN.3  W3.QUIEN.4
          W4.QUIEN.1  W4.QUIEN.2  W4.QUIEN.3  W4.QUIEN.4

    Retorna
    -------
    pd.DataFrame
        Copia del DataFrame original con la nueva columna 'W.QUIEN'.

    Lanza
    -----
    KeyError
        Si faltan columnas requeridas.
    """
    df_out = df.copy()

    # Columnas relacionadas con "padres" (.1 y .2) y "no padres" (.3 y .4)
    parent_cols = [
        f"W{i}.QUIEN.{j}"
        for i in range(1, 5)
        for j in (1, 2)
    ]
    non_parent_cols = [
        f"W{i}.QUIEN.{j}"
        for i in range(1, 5)
        for j in (3, 4)
    ]
    all_cols = parent_cols + non_parent_cols

    # Validar existencia de columnas
    missing = [c for c in all_cols if c not in df_out.columns]
    if missing:
        raise KeyError(f"Faltan las columnas requeridas: {missing}")

    # Flags por fila
    all_nan    = df_out[all_cols].isna().all(axis=1)
    has_parent = df_out[parent_cols].eq(1).any(axis=1)
    has_nonpar = df_out[non_parent_cols].eq(1).any(axis=1)

    # Inicializamos con 0 ("No padres")
    df_out["W.QUIEN"] = 0

    # 2 = Ambos (tienen al menos un 1 en ambos grupos)
    df_out.loc[has_parent & has_nonpar, "W.QUIEN"] = 2
    # 1 = Padres únicamente
    df_out.loc[has_parent & ~has_nonpar, "W.QUIEN"] = 1
    # 0 = No padres únicamente (ya es 0 por defecto)
    # 3 = Nadie (todas NaN)
    df_out.loc[all_nan, "W.QUIEN"] = 3

    return df_out

def transform_PAIS(df: pd.DataFrame) -> pd.DataFrame:
    """
    Imputa los valores faltantes en la columna 'PAÍS' con el valor 2.

    Pasos:
      1. Verifica que exista la columna 'PAÍS'.
      2. Rellena los NaN de 'PAÍS' con 2.
      3. Convierte la columna resultante a entero si procede.

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame que debe contener la columna 'PAÍS'.

    Retorna
    -------
    pd.DataFrame
        Copia del DataFrame original con la columna 'PAÍS' imputada.

    Lanza
    -----
    KeyError
        Si falta la columna 'PAÍS'.
    """
    # Copiar para no modificar el original
    df_transformed = df.copy()

    # Verificar existencia de la columna
    if "PAÍS" not in df_transformed.columns:
        raise KeyError("Falta la columna requerida: 'PAÍS'")

    # Imputar NaN con 2 y convertir a entero
    df_transformed["PAÍS"] = df_transformed["PAÍS"].fillna(2).astype(int)

    return df_transformed

def transform_MORAL(df: pd.DataFrame) -> pd.DataFrame:
    """
    Imputa valores faltantes en las columnas 'MORAL1' a 'MORAL5' con la media de cada columna,
    y añade dos nuevas columnas:
      - 'MORAL.MEAN': media por fila de los cinco ítems imputados.
      - 'MORAL.VAR' : varianza por fila de los cinco ítems imputados (varianza poblacional, ddof=0).

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame de entrada que debe contener las columnas
        'MORAL1', 'MORAL2', 'MORAL3', 'MORAL4', 'MORAL5'.

    Retorna
    -------
    pd.DataFrame
        Copia del DataFrame original con:
        - Las columnas 'MORAL1'...'MORAL5' imputadas sin NaN.
        - Dos columnas nuevas: 'MORAL.MEAN' y 'MORAL.VAR'.

    Lanza
    -----
    KeyError
        Si falta alguna de las columnas requeridas.
    """
    # Trabajar sobre copia para no mutar el df original
    df_transformed = df.copy()
    cols = [f"MORAL{i}" for i in range(1, 6)]

    # Verificar existencia de columnas
    missing = [c for c in cols if c not in df_transformed.columns]
    if missing:
        raise KeyError(f"Faltan las columnas requeridas: {missing}")

    # Imputación: llenar NaN con la media de cada columna
    for col in cols:
        mean_val = df_transformed[col].mean(skipna=True)
        df_transformed[col] = df_transformed[col].fillna(mean_val)

    # Cálculo de la media y varianza por fila
    df_transformed["MORAL.MEAN"] = df_transformed[cols].mean(axis=1)
    df_transformed["MORAL.VAR"]  = df_transformed[cols].var(axis=1, ddof=0)

    return df_transformed

def transform_PEXPCONTACT(df: pd.DataFrame) -> pd.DataFrame:
    """
    Crea la columna 'PEXP.CONTACT' codificada según el tipo de contacto presencial
    o virtual en la única serie PEXP1.CONTACT.{1..12}.

    Codificación de 'PEXP.CONTACT':
      - 3 : nadie       (todas las columnas PEXP1.CONTACT.1–12 son NaN)
      - 1 : presencial  (al menos un 1 en PEXP1.CONTACT.1–3, y ninguno en PEXP1.CONTACT.4–12)
      - 0 : virtual     (al menos un 1 en PEXP1.CONTACT.4–12, y ninguno en PEXP1.CONTACT.1–3)
      - 2 : ambos       (al menos un 1 en PEXP1.CONTACT.1–3 y al menos un 1 en PEXP1.CONTACT.4–12)

    Pasos
    -----
      1. Verifica que existan las columnas PEXP1.CONTACT.1 … PEXP1.CONTACT.12.
      2. Calcula:
         - `has_pres` = alguna de PEXP1.CONTACT.1–3 == 1
         - `has_virt` = alguna de PEXP1.CONTACT.4–12 == 1
         - `all_nan`  = todas PEXP1.CONTACT.1–12 son NaN
      3. Inicializa 'PEXP.CONTACT' = 0 (virtual).
      4. Asigna 2 donde `has_pres & has_virt` (ambos).
      5. Asigna 1 donde `has_pres & ~has_virt` (presencial únicamente).
      6. Asigna 3 donde `all_nan` (nadie).

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame que contiene las columnas PEXP1.CONTACT.1 … PEXP1.CONTACT.12.

    Retorna
    -------
    pd.DataFrame
        Copia del DataFrame original con la nueva columna 'PEXP.CONTACT'.

    Lanza
    -----
    KeyError
        Si falta alguna de las columnas esperadas.
    """
    # Copiar para no mutar el original
    df_transformed = df.copy()

    # Definir listas de columnas
    pres_cols = [f"PEXP1.CONTACT.{i}" for i in (1, 2, 3)]
    virt_cols = [f"PEXP1.CONTACT.{i}" for i in range(4, 13)]
    all_cols = pres_cols + virt_cols

    # Verificar existencia de todas las columnas
    missing = [c for c in all_cols if c not in df_transformed.columns]
    if missing:
        raise KeyError(f"Faltan las columnas requeridas: {missing}")

    # Flags por fila
    has_pres = df_transformed[pres_cols].eq(1).any(axis=1)
    has_virt = df_transformed[virt_cols].eq(1).any(axis=1)
    all_nan  = df_transformed[all_cols].isna().all(axis=1)

    # Inicializar como 'virtual' (0)
    df_transformed["PEXP.CONTACT"] = 0

    # Ambos (2)
    df_transformed.loc[has_pres & has_virt, "PEXP.CONTACT"] = 2
    # Presencial únicamente (1)
    df_transformed.loc[has_pres & ~has_virt, "PEXP.CONTACT"] = 1
    # Nadie (3)
    df_transformed.loc[all_nan, "PEXP.CONTACT"] = 3

    return df_transformed

def transform_VSQUIEN(df: pd.DataFrame) -> pd.DataFrame:
    """
    Crea la columna 'VS.QUIEN' codificada según presencia de respuestas en las
    columnas VS1.QUIEN.1–7, VS2.QUIEN.1–7, VS5.QUIEN.1–4 y VS6.QUIEN.1–4.

    Codificación de 'VS.QUIEN':
      - 3 : nadie (todas las columnas son NaN)
      - 1 : familia directa (al menos un 1 en VS1.QUIEN.1–2, VS2.QUIEN.1–2,
            VS5.QUIEN.2 o VS6.QUIEN.2, y ningún 1 fuera de ese grupo)
      - 0 : persona lejana (al menos un 1 en VS1.QUIEN.3–7, VS2.QUIEN.3–7,
            VS5.QUIEN.1,3–4 o VS6.QUIEN.1,3–4, y ningún 1 en el grupo 'familia directa')
      - 2 : ambos (hay al menos un 1 en ambos grupos 'familia directa' y
            'persona lejana')

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame que contiene las columnas:
          VS1.QUIEN.1 … VS1.QUIEN.7,
          VS2.QUIEN.1 … VS2.QUIEN.7,
          VS5.QUIEN.1 … VS5.QUIEN.4,
          VS6.QUIEN.1 … VS6.QUIEN.4.

    Retorna
    -------
    pd.DataFrame
        Copia del DataFrame original con la nueva columna 'VS.QUIEN'.

    Lanza
    -----
    KeyError
        Si falta alguna de las columnas requeridas.
    """
    df_out = df.copy()

    # Definir columnas de los dos grupos
    direct_cols = [
        "VS1.QUIEN.1", "VS1.QUIEN.2",
        "VS2.QUIEN.1", "VS2.QUIEN.2",
        "VS5.QUIEN.2",
        "VS6.QUIEN.2"
    ]
    remote_cols = [
        "VS1.QUIEN.3", "VS1.QUIEN.4", "VS1.QUIEN.5", "VS1.QUIEN.6", "VS1.QUIEN.7",
        "VS2.QUIEN.3", "VS2.QUIEN.4", "VS2.QUIEN.5", "VS2.QUIEN.6", "VS2.QUIEN.7",
        "VS5.QUIEN.1", "VS5.QUIEN.3", "VS5.QUIEN.4",
        "VS6.QUIEN.1", "VS6.QUIEN.3", "VS6.QUIEN.4"
    ]
    all_cols = direct_cols + remote_cols

    # Verificar que existan todas las columnas
    missing = [c for c in all_cols if c not in df_out.columns]
    if missing:
        raise KeyError(f"Faltan las columnas requeridas: {missing}")

    # Flags por fila
    all_nan    = df_out[all_cols].isna().all(axis=1)
    has_direct = df_out[direct_cols].eq(1).any(axis=1)
    has_remote = df_out[remote_cols].eq(1).any(axis=1)

    # Inicializar todo como 'nadie' (3)
    df_out["VS.QUIEN"] = 3

    # Ambos (2)
    df_out.loc[has_direct & has_remote, "VS.QUIEN"] = 2
    # Familia directa únicamente (1)
    df_out.loc[has_direct & ~has_remote & ~all_nan, "VS.QUIEN"] = 1
    # Persona lejana únicamente (0)
    df_out.loc[~has_direct & has_remote & ~all_nan, "VS.QUIEN"] = 0
    # Los que quedan (sin 1 alguno ni all_nan) permanecerán como 3 (nadie)

    return df_out

def transform_VPBULL(df: pd.DataFrame) -> pd.DataFrame:
    """
    Añade la columna 'VP.BULL' codificada según la presencia de bullying en VP1.BULL o VP2.BULL:

      - 1: si alguna de las columnas 'VP1.BULL' o 'VP2.BULL' es igual a 1.
      - 0: en cualquier otro caso (incluye valores 0 o NaN en ambas columnas).

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame que debe contener las columnas 'VP1.BULL' y 'VP2.BULL'.

    Retorna
    -------
    pd.DataFrame
        Copia del DataFrame original con la nueva columna 'VP.BULL'.

    Lanza
    -----
    KeyError
        Si faltan las columnas requeridas.
    """
    # Trabajar sobre copia para no modificar el original
    df_transformed = df.copy()

    required = ["VP1.BULL", "VP2.BULL"]
    missing = [col for col in required if col not in df_transformed.columns]
    if missing:
        raise KeyError(f"Faltan las columnas requeridas: {missing}")

    # Crear VP.BULL: 1 si alguna es 1, sino 0 (NaN → False)
    df_transformed["VP.BULL"] = (
        df_transformed[required]
        .eq(1)
        .any(axis=1)
        .astype(int)
    )

    return df_transformed
```

# 📊 Visualización de Variables

---

## 🔹 `one_hot_encode(df, columns)`
- Aplica **codificación one-hot** a las columnas categóricas indicadas.
- Transforma cada categoría en una columna binaria separada (0/1).
- No produce una visualización directamente, pero **prepara los datos para análisis visual posterior.**

---

## 🔹 `clasificar_columnas_por_unicidad(df)`
- Clasifica las columnas según su número de valores únicos:
  - **Binarias**: exactamente 2 valores distintos.
  - **Pocas categorías**: entre 3 y 10 valores únicos *(comentado pero no implementado)*.
  - **Muchas categorías**: más de 2 valores únicos.
- No genera gráficos, pero **ayuda a seleccionar columnas adecuadas para visualización** posterior (por ejemplo, barplots o histogramas).

---

## 🔹 `plot_histograms(df, columns)`
- Muestra un **grid de histogramas** para las columnas numéricas especificadas.
- Cada histograma:
  - Muestra la **distribución de frecuencias** de los valores.
  - Ayuda a detectar: asimetría, outliers, valores extremos, o datos centrados.
- Parámetros personalizables:
  - Número de filas/columnas del grid.
  - Tamaño del gráfico.
  - Número de bins (intervalos).

---

## 🔹 `plot_binary_barplots(df, columns)`
- Muestra un **grid de gráficos de barras** para variables binarias (0/1).
- Cada barplot representa:
  - **Conteo de ceros y unos** en la variable.
- Ideal para:
  - Visualizar la **distribución de respuestas sí/no**.
  - Identificar **desequilibrios** en variables objetivo o predictores binarios.

---

## 🔹 `project_correlation_html(df, html_path)`
- Visualiza las **relaciones de correlación** entre variables numéricas usando una **proyección tridimensional (MDS)**:
  1. Calcula la **matriz de correlación**.
  2. Transforma la correlación en una **matriz de distancias**.
  3. Aplica **Multidimensional Scaling (MDS)** en 3D.
  4. Genera un gráfico interactivo con **Plotly**, guardado como `.html`.
- Útil para:
  - Detectar **grupos de variables correlacionadas**.
  - Explorar visualmente la **estructura interna** del dataset.

---

## 🔹 `plot_correlation_heatmap(df)`
- Genera un **heatmap (mapa de calor)** de la matriz de correlación entre variables numéricas.
- Cada celda muestra:
  - El **coeficiente de correlación** entre dos variables.
- Ejes representados con índices numéricos (no nombres).
- Ideal para:
  - Detectar **relaciones lineales**.
  - Identificar **multicolinealidad** o variables redundantes.
- Requiere `matplotlib` y puede ajustarse el tamaño de texto.

---

## 📌 Nota General
Estas funciones cubren una gama completa de visualización para el análisis exploratorio de datos:
- Distribución univariada: `plot_histograms`, `plot_binary_barplots`.
- Relaciones multivariadas: `plot_correlation_heatmap`, `project_correlation_html`.
- Preprocesamiento útil para visualización: `one_hot_encode`, `clasificar_columnas_por_unicidad`.

```python
def one_hot_encode(df: pd.DataFrame, columns: list[str]) -> pd.DataFrame:
    """
    Aplica one-hot encoding a las columnas indicadas de un DataFrame.

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame original.
    columns : list[str]
        Lista con los nombres de las columnas a dummyficar.

    Devuelve
    -------
    pd.DataFrame
        Nuevo DataFrame con las columnas originales removidas y reemplazadas
        por sus variables dummy.
    """
    # Validación de que las columnas existen
    missing = [col for col in columns if col not in df.columns]
    if missing:
        raise KeyError(f"Las siguientes columnas no existen en el DataFrame: {missing}")

    # Usamos pandas.get_dummies para todo el DataFrame
    df_encoded = pd.get_dummies(df, columns=columns, dtype=int)

    return df_encoded
```

```python
def clasificar_columnas_por_unicidad(df):
    binarias = []        # Solo 2 valores únicos
    pocas_categorias = []  # Entre 3 y 10 valores únicos
    muchas_categorias = [] # Más de 10 valores únicos

    for col in df.columns:
        n_unique = df[col].nunique()
        if n_unique == 2:
            binarias.append(col)
        elif n_unique > 2:
            muchas_categorias.append(col)

    return binarias, muchas_categorias
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from typing import List, Tuple

def plot_histograms(
    df: pd.DataFrame,
    columns: List[str],
    nrows: int = 7,
    ncols: int = 5,
    figsize: Tuple[float,float] = (15, 21),
    bins: int = 10
) -> plt.Figure:
    """
    Dibuja un grid de histogramas para las columnas indicadas.

    Parámetros
    ----------
    df : pd.DataFrame
        El DataFrame con los datos.
    columns : List[str]
        Lista de nombres de columnas numéricas a graficar.
    nrows : int
        Número de filas de subplots (por defecto 7).
    ncols : int
        Número de columnas de subplots (por defecto 5).
    figsize : Tuple[float,float]
        Tamaño de la figura en pulgadas (ancho, alto).
    bins : int
        Número de bins para los histogramas.

    Retorna
    -------
    fig : matplotlib.figure.Figure
        La figura con todos los histogramas.
    """
    total_plots = nrows * ncols
    if len(columns) > total_plots:
        raise ValueError(f"Has pedido {len(columns)} histogramas, pero sólo hay espacio para {total_plots} subplots.")

    fig, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=figsize)
    axes = axes.flatten()  # simplifica el indexado

    for idx, col in enumerate(columns):
        if col not in df.columns:
            raise KeyError(f"La columna '{col}' no existe en el DataFrame.")
        ax = axes[idx]
        # Histograma con ejes automáticos
        df[col].dropna().hist(bins=bins, ax=ax)
        ax.set_title(col)
        ax.grid(False)

    # Ocultar los ejes sobrantes
    for j in range(len(columns), total_plots):
        axes[j].set_visible(False)

    fig.tight_layout()
    return fig
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from typing import List, Tuple

def plot_binary_barplots(
    df: pd.DataFrame,
    columns: List[str],
    nrows: int = 7,
    ncols: int = 6,
    figsize: Tuple[float, float] = (18, 21)
) -> plt.Figure:
    """
    Dibuja un grid de barplots de conteo para variables binarias.

    Parámetros
    ----------
    df : pd.DataFrame
        El DataFrame con las variables binarias (valores 0/1).
    columns : List[str]
        Lista de nombres de columnas a graficar (como máximo nrows*ncols).
    nrows : int
        Número de filas de subplots (por defecto 7).
    ncols : int
        Número de columnas de subplots (por defecto 6).
    figsize : Tuple[float, float]
        Tamaño de la figura en pulgadas (ancho, alto).

    Retorna
    -------
    fig : matplotlib.figure.Figure
        La figura con todos los barplots.
    """
    total_plots = nrows * ncols
    if len(columns) > total_plots:
        raise ValueError(
            f"Has pedido {len(columns)} gráficos, pero sólo hay espacio para {total_plots} subplots."
        )

    fig, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=figsize)
    axes = axes.flatten()

    for idx, col in enumerate(columns):
        if col not in df.columns:
            raise KeyError(f"La columna '{col}' no existe en el DataFrame.")
        ax = axes[idx]

        # Cuenta valores 0 y 1 (incluye ceros si faltan)
        counts = df[col].value_counts(dropna=False).reindex([0,1], fill_value=0)
        ax.bar(counts.index.astype(str), counts.values)
        ax.set_title(col)
        ax.set_xlabel('')
        ax.set_ylabel('Count')
        ax.set_ylim(0, counts.values.max() * 1.05)  # un pequeño margen arriba
        ax.grid(axis='y', linestyle='--', alpha=0.5)

    # Ocultar ejes sobrantes
    for j in range(len(columns), total_plots):
        axes[j].set_visible(False)

    fig.tight_layout()
    return fig
```

```python
import numpy as np
import pandas as pd
from sklearn.manifold import MDS
import plotly.express as px

def project_correlation_html(
    df: pd.DataFrame,
    html_path: str = '/mnt/data/correlation_mds.html',
    random_state: int = 42
) -> pd.DataFrame:
    """
    1) Calcula la matriz de correlación de df.
    2) La convierte en distancias: d_{ij} = sqrt(0.5*(1 - r_{ij})).
    3) Aplica MDS clásico para proyectar en 3 dimensiones.
    4) Crea un scatter_3d interactivo con Plotly y lo guarda en html_path.

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame con variables numéricas.
    html_path : str
        Ruta donde se guardará el fichero HTML interactivo.
    random_state : int
        Semilla para reproducibilidad de MDS.

    Devuelve
    -------
    coords_df : pd.DataFrame
        DataFrame con columnas ['x', 'y', 'z'] y como índice los nombres de las variables.
    """
    # 1. Correlación
    corr = df.corr()

    # 2. De correlación a distancia
    dist = np.sqrt(0.5 * (1 - corr.values))

    # 3. MDS clásico
    mds = MDS(
        n_components=3,
        dissimilarity='precomputed',
        random_state=random_state
    )
    coords = mds.fit_transform(dist)

    # 4. DataFrame de coordenadas
    coords_df = pd.DataFrame(
        coords,
        index=corr.index,
        columns=['x', 'y', 'z']
    )

    # 5. Plotly 3D scatter
    df_plot = coords_df.reset_index().rename(columns={'index':'variable'})
    fig = px.scatter_3d(
        df_plot,
        x='x', y='y', z='z',
        text='variable',
        title='Proyección 3D de variables según su correlación'
    )
    fig.update_traces(textposition='top center')
    fig.write_html(html_path, include_plotlyjs='cdn')

    print(f"✅ Gráfico interactivo guardado en: {html_path}")
    return coords_df
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

def plot_correlation_heatmap(df, label_fontsize=8):
    """
    Genera un heatmap de la matriz de correlación de df.
    En los ejes muestra el índice numérico de cada columna,
    con el tamaño de letra de las etiquetas ajustable.
    Devuelve la figura de matplotlib que contiene el heatmap.
    """
    corr = df.corr()
    fig, ax = plt.subplots()
    cax = ax.imshow(corr, interpolation='nearest')
    fig.colorbar(cax, ax=ax)

    n = len(corr.columns)
    # Usamos números de columna en lugar de nombres
    ax.set_xticks(range(n))
    ax.set_xticklabels(range(n), rotation=90, fontsize=label_fontsize)
    ax.set_yticks(range(n))
    ax.set_yticklabels(range(n), fontsize=label_fontsize)

    plt.tight_layout()
    return fig

fig = plot_correlation_heatmap(df_feature)
plt.show()
```

```python
from typing import Any, Dict, List, Tuple

def encode_binary_columns(
    df: pd.DataFrame
) -> Tuple[pd.DataFrame, Dict[str, List[Tuple[Any, int]]]]:
    """
    Convierte en un DataFrame todas las columnas binarias (exactamente 2 valores únicos)
    a 0/1 asignando 1 al valor más frecuente y 0 al otro. Luego imputa los NaN
    con el valor más común en cada columna. Devuelve el DataFrame resultante y un
    diccionario con la asignación por columna.

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame original con columnas de cualquier tipo.

    Devuelve
    -------
    df_encoded : pd.DataFrame
        DataFrame con las columnas binarias codificadas a 0/1 e imputadas.
    mapping_dict : Dict[str, List[Tuple[Any,int]]]
        Diccionario donde cada clave es el nombre de una columna binaria y
        su valor es una lista de dos tuplas (valor_original, valor_binario).
    """
    df_encoded = df.copy()
    mapping_dict: Dict[str, List[Tuple[Any, int]]] = {}

    for col in df_encoded.columns:
        # valores únicos (sin contar NaN)
        uniques = df_encoded[col].dropna().unique()
        if len(uniques) != 2:
            continue  # no es columna binaria

        # cuál es el valor más frecuente (modo)
        counts = df_encoded[col].value_counts(dropna=True)
        most_common = counts.idxmax()

        # el otro valor
        other = [v for v in uniques if v != most_common][0]

        # creamos el mapping: other→0, most_common→1
        mapping = {other: 0, most_common: 1}

        # aplicamos el mapeo y luego imputamos NaN con el valor más común (1)
        df_encoded[col] = df_encoded[col].map(mapping)
        df_encoded[col].fillna(mapping[most_common], inplace=True)

        # guardamos la lista de tuplas en el diccionario
        # (ordenada para que primero vaya el 0 y luego el 1)
        mapping_list = sorted(mapping.items(), key=lambda x: x[1])
        mapping_dict[col] = mapping_list

    return df_encoded, mapping_dict

def one_hot_no_impute(df: pd.DataFrame) -> pd.DataFrame:
    """
    Hace one-hot encoding de TODO el DataFrame:
      - Primero convierte cada columna a 'category',
        para que get_dummies lo trate como categórico.
      - Luego aplica pd.get_dummies con dummy_na=False,
        de forma que las filas con NaN queden con todos 0
        en las dummies de esa variable.

    Parámetros
    ----------
    df : pd.DataFrame
        DataFrame con variables categóricas (incluso si están como int/float).

    Devuelve
    -------
    pd.DataFrame
        Sólo columnas binarias (0/1), sin NaN.
    """
    # convertimos todo a 'category'
    df_cat = df.copy().astype('category')
    # one-hot sin generar columna para NaN
    df_ohe = pd.get_dummies(df_cat, dummy_na=False)
    return df_ohe
```

# 🧾 Conclusiones del Análisis Univariante y Segmentación de Variables

Tras un exhaustivo análisis univariante, se ha procedido a organizar y clasificar todas las variables del dataset en **cuatro bloques funcionales**. Esta segmentación ha permitido aclarar el papel que desempeña cada conjunto de variables en el proceso analítico y de modelado. A continuación se resumen los hallazgos y la lógica detrás de estas decisiones.

---

## 🔹 1. Variables Globales

### 📌 Características
- Variables **estables y estructurales** (como edad, etnia, país, convivencia, moralidad, impulsividad, etc.).
- No presentan **desbalances importantes**, ni valores extremos frecuentes, ni alta cantidad de `NaNs`.
- Tienen un **contenido descriptivo general** del individuo, y pueden funcionar como **buenos predictores base**.

### 🔍 Propuesta de uso
- Se utilizarán como **features predictivas directas** en modelos de clasificación.
- Además, se plantea aplicar **técnicas de clustering** (e.g., K-Means, HDBSCAN, GMM) exclusivamente sobre estas variables para:
  - Capturar **patrones latentes** o perfiles poblacionales homogéneos.
  - Usar el clúster resultante como una **variable adicional enriquecida**.

### 🤝 Vinculación cruzada con otros bloques
- Los clústeres definidos sobre estas variables globales se usarán para analizar:
  - Cómo se distribuyen en ellos las **víctimas** y los **perpetradores**.
  - Si existen **tipologías conjuntas**, por ejemplo:
    - Clústeres con alta densidad de víctimas que también tienen tendencia a ser perpetradores.
    - Clústeres sin involucramiento en violencia, etc.

---

## 🔸 2. Variables de Víctima

### 📌 Características
- Definen eventos de victimización.
- En su mayoría son **variables binarias o discretas muy desbalanceadas**.
  - Muchas de ellas presentan distribución con gran mayoría en la categoría 0.
- Tienen un alto contenido informativo, pero deben manejarse con **cuidado analítico**:
  - Preprocesamiento adecuado (p. ej., binarización, oversampling).
  - Evaluación específica de su contribución predictiva.

### 🎯 Uso
- Se utilizarán:
  - Para construir los **targets de victimización** (`VÍCTIMA`, `POLIVICTIMIZACION`, etc.).
  - Como grupo de variables que podrían usarse para estudiar **subtipos de víctimas**.

## 🔸 3. Variables de Perpetrador

### 📌 Características
- Siguen una lógica estructural muy similar a las de víctima:
  - Alto desbalance.
  - Frecuentemente discretas con conteos acumulados.
- Son relevantes para definir el comportamiento de agresión o transgresión.

### 🎯 Uso
- Fundamentales para construir los **targets de perpetración** (`PERPETRADOR`, `POLIPERPETRACION`, etc.).
- Permiten explorar perfiles de **riesgo de conducta agresiva**, e incluso prever tipos de comportamiento violento.

## ❌ 4. Variables Useless

### 📌 Características
- Variables extremadamente específicas, derivadas o redundantes.
- Incluyen:
  - Versiones preliminares de variables luego transformadas (`X1`, `X2`, `X.BN`, `X.SUM`, etc.).
  - Preguntas abiertas, etiquetas categóricas extensas o con baja frecuencia.
  - Codificaciones intermedias, inputs crudos y variables administrativas (e.g., `FECHAENCUESTA`, `GENERO.OTRO`, etc.).

### 🚫 Decisión
- Se han **eliminado del dataset final**.
- Su inclusión en modelos o análisis sólo añadiría ruido o colinealidad.
- No aportan valor informativo adicional ni son generalizables.

## 🧠 Consideraciones Finales

- Se han creado **transformaciones robustas** para resumir escalas (`IMPULS`, `AUTOEFIC`, `APOYO`, `MORAL`...), manteniendo la información relevante en forma agregada.
- Se ha conservado una estructura limpia de **features clave**, con un enfoque claro hacia:
  - Modelado predictivo.
  - Exploración de perfiles latentes (via clustering).
  - Interpretación de resultados en función de la interacción víctima-perpetrador.
